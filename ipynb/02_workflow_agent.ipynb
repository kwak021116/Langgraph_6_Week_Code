## 1. Workflow: 코드의 조건대로 계산

PPT와 같은 a=10, b=20을 사용합니다.

```python
a = 10
b = 20

if a >= 10:
    result = a + b
else:
    result = a * b

print(result)
```

**결과 해설**

`30`입니다. `a >= 10`이라는 규칙을 사람이 정했습니다.

## 2. 모델이 사용할 함수

PPT에 있는 덧셈과 곱셈 함수입니다.

```python
def add_numbers(a: int, b: int) -> int:
    """두 정수 a와 b를 더합니다."""
    return a + b


def multiply_numbers(a: int, b: int) -> int:
    """두 정수 a와 b를 곱합니다."""
    return a * b
```

## 3. 모델에 도구 연결

키와 모델명은 `.env`에서 읽습니다.

```python
from langchain_core.messages import HumanMessage

llm = make_llm()
tools = [add_numbers, multiply_numbers]
tool_llm = llm.bind_tools(tools)
```

## 4. 덧셈을 요청

어느 도구와 인자를 요청하는지 확인합니다. 여기서는 함수를 실행하지 않습니다.

```python
message = HumanMessage(content="10과 20을 더해줘. 덧셈 도구를 사용해줘.")
response = tool_llm.invoke([message])

response.pretty_print()
```

**결과 해설**

도구 요청이 있다면 add_numbers와 a=10, b=20을 확인합니다. 이 예시는 Agent의 도구 선택 부분이며 반복 실행은 11쪽에서 만듭니다.
