## 1. 메시지는 누가 한 말인지 담습니다

단순한 인사로 역할을 확인합니다. 아래는 역할 설명용 객체이며 API 응답이 아닙니다.

```python
from langchain_core.messages import HumanMessage, AIMessage

user_message = HumanMessage(content="안녕")
ai_message = AIMessage(content="안녕하세요!")

print("사용자:", user_message.content)
print("모델:", ai_message.content)
```

## A-1. operator.add 입력

메시지 ID가 모두 1입니다. 같은 말을 수정하고 싶다고 생각해보세요.

```python
import operator

old_messages = [
    HumanMessage(content="안녕", id="1"),
]

new_messages = [
    HumanMessage(content="안녕하세요", id="1"),
]
```

## A-2. operator.add 결과

기존 메시지와 새 메시지를 그대로 이어 붙입니다.

```python
result = operator.add(old_messages, new_messages)

for message in result:
    print(message.id, message.content)

print("메시지 개수:", len(result))
```

**결과 해설**

`1 안녕`, `1 안녕하세요`가 모두 출력됩니다. 메시지는 2개입니다.

## B-1. add_messages 입력

A와 똑같은 메시지로 다시 시작합니다.

```python
from langgraph.graph.message import add_messages

old_messages = [
    HumanMessage(content="안녕", id="1"),
]

new_messages = [
    HumanMessage(content="안녕하세요", id="1"),
]
```

## B-2. add_messages 결과

같은 ID를 찾아 새 내용으로 바꿉니다.

```python
result = add_messages(old_messages, new_messages)

for message in result:
    print(message.id, message.content)

print("메시지 개수:", len(result))
```

**결과 해설**

`1 안녕하세요` 하나만 남습니다. 메시지는 1개입니다.

## C. ID가 다르면 새 메시지 추가

이번에는 두 번째 메시지 ID가 2입니다.

```python
old_messages = [
    HumanMessage(content="안녕", id="1"),
]

new_messages = [
    HumanMessage(content="반가워", id="2"),
]

result = add_messages(old_messages, new_messages)

for message in result:
    print(message.id, message.content)
```

**결과 해설**

`1 안녕`, `2 반가워`가 남습니다.
